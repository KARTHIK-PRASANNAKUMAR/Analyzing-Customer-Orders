```python
#Store Customer Orders
```

```python
# List of customers
customers = ["Alice", "Bob", "Charlie", "Diana"]

# Orders: (customer, product, price, category)
orders = [
    ("Alice", "Laptop", 800, "Electronics"),
    ("Bob", "T-shirt", 20, "Clothing"),
    ("Alice", "Headphones", 50, "Electronics"),
    ("Charlie", "Sofa", 300, "Home Essentials"),
    ("Diana", "Dress", 70, "Clothing"),
    ("Bob", "Blender", 60, "Home Essentials"),
    ("Charlie", "Phone", 500, "Electronics"),
]

# Dictionary: customer → list of products
customer_orders = {}

for customer, product, price, category in orders:
    customer_orders.setdefault(customer, []).append(product)

print(customer_orders)
```

```python
#Classify Products by Category
```

```python
# Product → category mapping
product_category = {}

for _, product, _, category in orders:
    product_category[product] = category

# Unique categories
categories = set(product_category.values())

print("Categories:", categories)
```

```python
#Analyze Customer Spending
```

```python
# Total spending per customer
customer_spending = {}

for customer, _, price, _ in orders:
    customer_spending[customer] = customer_spending.get(customer, 0) + price

# Classification
customer_classification = {}

for customer, total in customer_spending.items():
    if total > 100:
        customer_classification[customer] = "High-value"
    elif 50 <= total <= 100:
        customer_classification[customer] = "Moderate"
    else:
        customer_classification[customer] = "Low-value"

print(customer_classification)
```

```python
#Generate Business Insights
```

```python
category_revenue = {}

for _, _, price, category in orders:
    category_revenue[category] = category_revenue.get(category, 0) + price

print("Category Revenue: ",category_revenue)
```

```python
#Unique Products
```

```python
unique_products = {product for _, product, _, _ in orders}
print("Product Type:" ,unique_products)
```

```python
#Customers Who Bought Electronics
```

```python
electronics_customers = [
    customer for customer, _, _, category in orders
    if category == "Electronics"
]

print(""set(electronics_customers))
```

```python
#Top 3 Highest-Spending Customers
```

```python
top_customers = sorted(customer_spending.items(), key=lambda x: x[1], reverse=True)[:3]

print(top_customers)
```

```python
#Customers Buying Multiple Categories
```

```python
customer_categories = {}

for customer, _, _, category in orders:
    customer_categories.setdefault(customer, set()).add(category)

multi_category_customers = [c for c, cats in customer_categories.items() if len(cats) > 1]

print(multi_category_customers)
```

```python
#Customers Who Bought Both Electronics & Clothing
```

```python
customer_categories = {}

for customer, product, price, category in orders:
    customer_categories.setdefault(customer, set()).add(category)

both_categories = [
    customer
    for customer, categories in customer_categories.items()
    if "Electronics" in categories and "Clothing" in categories
]

if both_categories:
    print("Customers who bought both Electronics and Clothing:", both_categories)
    print(both_categories)
else:
    print("No customers bought both Electronics and Clothing.")
```
